```
# ====================================================================
# 🎬 YouTube Social Media Generator - IMPROVED & MODERNIZED VERSION
# ====================================================================
# This script uses yt-dlp to fetch video metadata and a HuggingFace T5 model
# (google/flan-t5-base) for generating platform-specific social media posts
# (LinkedIn, Facebook, Twitter) based on the video's title and description.

# --- 1. DEPENDENCY INSTALLATION ---
# Removed 'gradio-themes' as it caused an error in the original run.
!pip -q install yt-dlp transformers sentencepiece accelerate gradio nltk emoji --upgrade

import re, time, json, random
import yt_dlp
import nltk
import gradio as gr
import numpy as np
import emoji
from datetime import datetime

from transformers import pipeline

# Download NLTK data
nltk.download("punkt", quiet=True)
nltk.download("stopwords", quiet=True)

# -----------------------------------
# 2. LOAD NLP MODEL
# -----------------------------------
print("🔄 Loading AI model: Flan-T5-Base...")
# Define the maximum length for the generated text (e.g., summary)
MODEL_MAX_LENGTH = 200

try:
    generator = pipeline(
        "text2text-generation",
        model="google/flan-t5-base",
        max_length=MODEL_MAX_LENGTH,
        repetition_penalty=2.5,
        temperature=0.7,
        do_sample=True
    )
    print("✅ Model loaded successfully!")
except Exception as e:
    print(f"❌ Error loading model: {e}")
    # Fallback/stub if model loading fails
    generator = lambda prompt, max_length: [{"generated_text": "AI generation service unavailable. Using fallback summary."}]


# -----------------------------------
# 3. EMOJI AND TEMPLATE DATABASE
# -----------------------------------
PLATFORM_TEMPLATES = {
    "LinkedIn": {
        "emoji": ["💼", "👔", "📊", "💡", "🚀", "🎯", "🌟"],
        "style": "professional and informative",
        "hashtags": ["#CareerGrowth", "#ProfessionalDevelopment", "#Learning", "#Skills", "#Education"]
    },
    "Facebook": {
        "emoji": ["📘", "👍", "❤️", "😊", "🎬", "👥", "📱", "✨"],
        "style": "friendly and community-focused",
        "hashtags": ["#FacebookWatch", "#Share", "#Community", "#VideoOfTheDay", "#WatchTogether"]
    },
    "Twitter": {
        "emoji": ["🐦", "💬", "🔥", "🚀", "📈", "💎", "🎯", "⚡"],
        "style": "concise and viral-focused",
        "hashtags": ["#Thread", "#EduTwitter", "#LearnEveryday", "#YouTube", "#MustWatch"]
    }
}

# -----------------------------------
# 4. UTILITY FUNCTIONS
# -----------------------------------

def validate_youtube_url(url):
    """Validate and normalize a YouTube URL to a standard watch format."""
    patterns = [
        r'(?:youtube\.com\/(?:watch\?v=|embed\/|v\/)|youtu\.be\/)([a-zA-Z0-9_-]{11})'
    ]
    for pattern in patterns:
        match = re.search(pattern, url)
        if match:
            video_id = match.group(1)
            return f"https://www.youtube.com/watch?v={video_id}"
    return None

def extract_video_metadata(url):
    """Extract video metadata using yt-dlp with careful error handling."""
    try:
        ydl_opts = {
            'quiet': True,
            'no_warnings': True,
            'extract_flat': True,
            'force_generic_extractor': False,
            'sleep_interval': 1,
            'max_sleep_interval': 3,
        }

        with yt_dlp.YoutubeDL(ydl_opts) as ydl:
            info = ydl.extract_info(url, download=False)

            # Format duration
            duration_seconds = info.get('duration', 0)
            duration = "Unknown"
            if duration_seconds:
                minutes = duration_seconds // 60
                seconds = duration_seconds % 60
                duration = f"{minutes}:{seconds:02d}"

            # Get statistics
            view_count = info.get('view_count', 0)
            like_count = info.get('like_count', 0)

            # Format description (limit to first 2000 chars)
            description = info.get('description', '')
            if len(description) > 2000:
                description = description[:2000] + "..."

            return {
                "success": True,
                "title": info.get('title', 'Untitled Video'),
                "description": description,
                "channel": info.get('uploader', 'Unknown Channel'),
                "duration": duration,
                "views": f"{view_count:,}" if view_count else "Not available",
                "likes": f"{like_count:,}" if like_count else "Not available",
                "url": url,
                # Use a specific high-quality thumbnail if available
                "thumbnail": info.get('thumbnail', '') or info.get('thumbnails', [{}])[-1].get('url', '')
            }

    except Exception as e:
        print(f"Error extracting metadata: {e}")
        return {
            "success": False,
            "title": "Error Processing Video",
            "description": f"Could not extract video information. Please check the URL.",
            "channel": "Unknown",
            "duration": "0:00",
            "views": "0",
            "likes": "0",
            "url": url,
            "thumbnail": "https://placehold.co/480x270/FF0000/FFFFFF?text=Error"
        }

def analyze_video_content(title, description):
    """Analyze video content for categorization."""
    text = f"{title} {description}".lower()

    content_types = {
        "Educational 📚": ["tutorial", "lesson", "course", "learn", "education", "teaching", "skill"],
        "Tutorial 🎓": ["how to", "step by step", "guide", "walkthrough", "demonstration", "install"],
        "Podcast 🎙️": ["podcast", "interview", "discussion", "conversation", "talk show", "guest"],
        "Entertainment 🎬": ["trailer", "movie", "film", "music video", "performance", "funny", "reaction"],
        "News 📰": ["news", "update", "breaking", "current events", "report", "analysis"],
        "Gaming 🎮": ["gaming", "gameplay", "review", "esports", "stream"],
        "Vlog 🎥": ["vlog", "day in life", "daily vlog", "storytime", "personal"],
        "Review ⭐": ["review", "unboxing", "test", "comparison", "best"],
        "Cooking 👨‍🍳": ["recipe", "cooking", "food", "baking", "culinary", "eat"],
    }

    for content_type, keywords in content_types.items():
        if any(keyword in text for keyword in keywords):
            return content_type

    return "General Content 🎯"

def generate_summary(title, description, content_type, tone):
    """Generate a concise 3-sentence summary using the T5 model."""
    clean_desc = re.sub(r'\\s+', ' ', description[:1000]).strip()

    # Base prompt adjusted for tone and platform style
    prompt = f"Act as a content marketing specialist. Generate a {tone} 3-sentence summary for a social media post based on this video: '{title}'. The summary should be concise, compelling, and aimed at encouraging views. Highlight the main topic and one key takeaway.\n\nDescription: {clean_desc}"

    try:
        # T5 Model generates the text
        result = generator(prompt, max_length=MODEL_MAX_LENGTH)[0]["generated_text"]

        # Post-process: attempt to ensure exactly 3 sentences
        sentences = nltk.sent_tokenize(result)
        if len(sentences) > 3:
            sentences = sentences[:3]
        elif len(sentences) < 3:
            # Add a compelling call to action if needed
            while len(sentences) < 3:
                sentences.append("Don't miss out on these valuable insights!")

        return ' '.join(sentences)
    except:
        # Fallback summary
        return f"This video titled \"{title}\" is an engaging watch. It provides valuable information and key takeaways on the subject matter. Click the link to learn more and dive into the discussion!"

def extract_keywords(text, max_keywords=8):
    """Extract relevant keywords from text using frequency analysis."""
    try:
        # Clean text
        text = re.sub(r'[^a-zA-Z0-9\\s]', ' ', text.lower())

        # Remove stopwords
        stop_words = set(nltk.corpus.stopwords.words('english'))
        words = [word for word in text.split() if word not in stop_words and len(word) > 3]

        from collections import Counter
        word_freq = Counter(words)

        # Get most common words, ensure capitalization for better hashtags
        keywords = [word.title() for word, _ in word_freq.most_common(max_keywords) if word.lower() not in ["video", "content", "youtube"]]

        if len(keywords) < 3:
            keywords.extend(["Learn", "Knowledge", "Insights", "Future"])
            keywords = list(set(keywords)) # Deduplicate and limit

        return keywords[:max_keywords]
    except:
        return ["AI", "Education", "Learning", "DigitalMarketing", "VideoContent"]

def create_social_post(platform, video_data, summary, keywords):
    """Create platform-specific social media posts with optimized formatting."""
    config = PLATFORM_TEMPLATES[platform]
    emoji_char = random.choice(config["emoji"])

    # Prepare hashtags (5 keyword hashtags + 3 platform-specific)
    keyword_hashtags = [f"#{k.replace(' ', '')}" for k in keywords[:5]]
    all_hashtags = keyword_hashtags + config["hashtags"][:3]
    hashtag_text = " ".join(all_hashtags)

    # Clean summary parts for use in templates
    summary_parts = summary.split('.')
    s1 = summary_parts[0] + ('.' if len(summary_parts[0]) > 0 else '')
    s2 = (summary_parts[1].strip() + '.') if len(summary_parts) > 1 else 'Check it out!'

    # Platform-specific templates
    if platform == "LinkedIn":
        post = (
            f"{emoji_char} **Highly insightful content alert for my professional network!**\\n\\n"
            f"The latest video, *\\\"{video_data['title']}\\\"*, from {video_data['channel']} provides crucial updates on a key topic.\\n\\n"
            f"👉 {s1}\\n"
            f"📈 Key Takeaway: {s2}\\n\\n"
            f"🔗 Full Video ({video_data['duration']}): {video_data['url']}\\n\\n"
            f"Who else is working on this? Let me know your thoughts!\\n\\n"
            f"{hashtag_text}"
        )
    elif platform == "Facebook":
        post = (
            f"{emoji_char} Awesome find! Sharing this gem with the community.\\n\\n"
            f"***{video_data['title']}*** is a must-watch, perfect for a weekend viewing.\\n\\n"
            f"What's covered? {s1} {s2}\\n\\n"
            f"Join the discussion below!\\n"
            f"⬇️ Watch Now ({video_data['views']} views): {video_data['url']}\\n\\n"
            f"{hashtag_text}"
        )
    elif platform == "Twitter":
        # Twitter is optimized for brevity and link placement
        post_lines = [
            f"{emoji_char} Must-watch alert! 🚨",
            f"🔥 Video: \"{video_data['title'][:50]}...\"",
            f"Summary: {s1[:90]}...",
            f"⏱️ {video_data['duration']} | 👁️ {video_data['views']} views",
            f"🔗 {video_data['url']}",
            f"{hashtag_text}"
        ]
        post = "\n".join(post_lines)

    return post

# -----------------------------------
# 5. MAIN PROCESSING FUNCTION
# -----------------------------------
def process_youtube_video(url, tone):
    """Main function to process YouTube URL and generate all outputs."""

    # 1. Validate URL
    valid_url = validate_youtube_url(url)
    if not valid_url:
        error_msg = "❌ Invalid YouTube URL. Please check the format and try again."
        empty_outputs = [error_msg] * 3 + ["Invalid URL", "N/A", "N/A", error_msg, "N/A", "N/A", ""]
        # Ensure we return 10 items (3 posts + 6 analysis fields + 1 thumbnail)
        return empty_outputs + ["https://placehold.co/480x270/FF0000/FFFFFF?text=Error"]

    start_time = time.time()

    try:
        # 2. Extract metadata
        gr.Info(f"Fetching metadata for: {valid_url}")
        video_data = extract_video_metadata(valid_url)

        if not video_data["success"]:
            raise Exception(video_data["description"])

        # 3. Analyze content
        content_type = analyze_video_content(video_data["title"], video_data["description"])

        # 4. Generate summary
        summary = generate_summary(video_data["title"], video_data["description"], content_type, tone)

        # 5. Extract keywords
        analysis_text = f"{video_data['title']} {summary} {video_data['description']}"
        keywords = extract_keywords(analysis_text)

        # 6. Generate social media posts
        linkedin_post = create_social_post("LinkedIn", video_data, summary, keywords)
        facebook_post = create_social_post("Facebook", video_data, summary, keywords)
        twitter_post = create_social_post("Twitter", video_data, summary, keywords)

        # 7. Prepare analysis data
        processing_time = f"{time.time() - start_time:.2f} seconds"
        stats = f"Views: {video_data['views']} | Likes: {video_data['likes']} | Duration: {video_data['duration']}"

        gr.Info("✅ Content generated successfully!")

        return [
            linkedin_post,
            facebook_post,
            twitter_post,
            video_data["title"],
            video_data["channel"],
            content_type,
            summary,
            ", ".join(keywords),
            stats,
            video_data["thumbnail"]
        ]

    except Exception as e:
        error_msg = f"⚠️ Processing Error: {str(e)[:100]}..."
        print(f"Full Error: {e}")
        error_summary = f"Failed to process video: {str(e)}"
        empty_outputs = [error_msg] * 3 + ["Error", "N/A", "N/A", error_summary, "N/A", "Error", ""]
        # Ensure we return 10 items (3 posts + 6 analysis fields + 1 thumbnail)
        return empty_outputs + ["https://placehold.co/480x270/FF0000/FFFFFF?text=Error"]

# -----------------------------------
# 6. MODERNIZED GRADIO INTERFACE
# -----------------------------------
def create_interface():
    """Create the modernized Gradio interface with custom CSS."""

    # Define the custom CSS for a cleaner, modern look
    custom_css = """
        /* Global Styling */
        body {
            font-family: 'Inter', sans-serif;
            background-color: #f7f7f7;
        }
        .gradio-container {
            max-width: 1300px;
            margin: auto;
            border-radius: 12px;
            box-shadow: 0 4px 20px rgba(0, 0, 0, 0.05);
            padding: 0 !important;
        }

        /* Header Styling */
        .title-header {
            text-align: center;
            background: linear-gradient(135deg, #1A73E8, #4285F4); /* Google Blue Gradient */
            padding: 30px 20px;
            border-top-left-radius: 12px;
            border-top-right-radius: 12px;
            color: white;
        }
        .title-header h1 {
            font-size: 2.2em;
            font-weight: 700;
            margin-bottom: 5px;
        }
        .title-header p {
            font-size: 1.1em;
            opacity: 0.9;
        }

        /* Input and Button Styling */
        .input-area {
            padding: 20px;
        }
        .generate-btn {
            height: 100%;
            min-height: 50px;
            font-size: 1.1em !important;
            font-weight: 600 !important;
            transition: all 0.3s ease;
        }
        .generate-btn:hover {
            box-shadow: 0 4px 10px rgba(66, 133, 244, 0.4);
            transform: translateY(-1px);
        }

        /* Output Box Styling */
        .output-box-container textarea {
            min-height: 150px !important;
            border-radius: 8px !important;
            border: 1px solid #ddd !important;
            background-color: #fff !important;
            padding: 15px;
            font-family: monospace; /* For clear text output */
        }

        /* Analysis Tab Layout */
        .analysis-section {
            padding: 20px;
        }
        .video-card {
            border: 1px solid #e0e0e0;
            border-radius: 10px;
            padding: 15px;
            margin-bottom: 20px;
            background-color: #fff;
            box-shadow: 0 2px 5px rgba(0,0,0,0.03);
        }
        /* Thumbnail aspect ratio fix for Gradio Image component */
        .thumbnail-container img {
            max-width: 100%;
            height: auto;
            border-radius: 8px;
        }
    """

    # Define outputs list globally for cleaner click event
    output_components = []

    with gr.Blocks(
        title="YouTube Social Media Generator",
        theme=gr.themes.Monochrome(), # Modern, clean theme
        css=custom_css
    ) as app:

        # --- Header ---
        gr.HTML(
            """
            <div class="title-header">
                <h1>🎬 YouTube Content Booster</h1>
                <p>Instantly generate perfectly formatted social media posts from any YouTube link using AI.</p>
            </div>
            """
        )

        # --- Input Section ---
        with gr.Row(elem_classes="input-area"):
            with gr.Column(scale=4):
                url_input = gr.Textbox(
                    label="📌 YouTube Video URL",
                    placeholder="Paste the YouTube link (e.g., https://youtu.be/dQw4w9WgXcQ)...",
                    lines=1
                )

                # Tone control for better summary generation
                tone_input = gr.Dropdown(
                    label="🗣️ Desired Post Tone",
                    choices=["Professional", "Friendly", "Casual", "Academic"],
                    value="Professional",
                    interactive=True
                )

                # Example URLs
                with gr.Accordion("📋 Examples & Tips", open=False):
                    gr.Markdown(
                        """
                        - **Tip:** The AI works best with videos that have detailed titles and descriptions.
                        - **Example Educational:** `https://youtu.be/B1J6Ou4q8vE`
                        - **Example Tutorial:** `https://youtu.be/UB1O30fR-EE`
                        """
                    )

            with gr.Column(scale=1):
                generate_btn = gr.Button(
                    "🚀 Generate Content",
                    variant="primary",
                    elem_classes="generate-btn"
                )

        # --- Output Tabs ---
        with gr.Tabs():

            # --- Platform Tabs (for Social Posts) ---
            with gr.Tab("💼 LinkedIn (Professional)"):
                gr.Markdown("### Professional & Informative Post")
                linkedin_output = gr.Textbox(
                    label="Generated LinkedIn Post (Optimized for professional audience)",
                    lines=10,
                    elem_classes="output-box-container",
                    # show_copy_button=True # Removed to fix TypeError
                )
                output_components.append(linkedin_output)

            with gr.Tab("📘 Facebook (Engaging)"):
                gr.Markdown("### Engaging & Community-Focused Post")
                facebook_output = gr.Textbox(
                    label="Generated Facebook Post (Great for groups and community sharing)",
                    lines=10,
                    elem_classes="output-box-container",
                    # show_copy_button=True # Removed to fix TypeError
                )
                output_components.append(facebook_output)

            with gr.Tab("🐦 Twitter (Concise)"):
                gr.Markdown("### Concise Tweet (Character Limited)")
                twitter_output = gr.Textbox(
                    label="Generated Tweet (Optimized for brevity and impact)",
                    lines=8,
                    elem_classes="output-box-container",
                    # show_copy_button=True # Removed to fix TypeError
                )
                output_components.append(twitter_output)

            # --- Analysis Tab ---
            with gr.Tab("📊 Video Analysis & Summary"):
                gr.Markdown("### Comprehensive Video Insights")

                with gr.Row(elem_classes="analysis-section"):
                    with gr.Column(scale=1):
                        # Video Thumbnail Display
                        thumbnail_output = gr.Image(
                            label="Video Thumbnail",
                            show_label=True,
                            interactive=False,
                            height=200,
                            elem_classes="thumbnail-container"
                        )
                        output_components.append(thumbnail_output)

                        # Basic Stats
                        stats_output = gr.Textbox(label="📈 Key Statistics", interactive=False)
                        output_components.append(stats_output)

                    with gr.Column(scale=2, elem_classes="video-card"):
                        title_output = gr.Textbox(label="🎬 Video Title", interactive=False, lines=2)
                        output_components.append(title_output)

                        channel_output = gr.Textbox(label="👤 Channel Name", interactive=False)
                        output_components.append(channel_output)

                        type_output = gr.Textbox(label="📊 Content Type Classification", interactive=False)
                        output_components.append(type_output)

                        # Keywords as a friendly list
                        keywords_output = gr.Textbox(
                            label="🏷️ Top Keywords (Use as additional hashtags or tags)",
                            lines=2,
                            interactive=False,
                            # show_copy_button=True # Removed to fix TypeError
                        )
                        output_components.append(keywords_output)


                summary_output = gr.Textbox(
                    label="📝 AI Generated Summary",
                    lines=5,
                    interactive=False,
                    # show_copy_button=True # Removed to fix TypeError
                )
                output_components.append(summary_output)

        # --- Connect Button to Function ---
        # The list of output components MUST match the order of returns in process_youtube_video
        # The correct order is: linkedin, facebook, twitter, title, channel, type, summary, keywords, stats, thumbnail
        generate_btn.click(
            fn=process_youtube_video,
            inputs=[url_input, tone_input],
            outputs=[
                linkedin_output,
                facebook_output,
                twitter_output,
                title_output,
                channel_output,
                type_output,
                summary_output,
                keywords_output,
                stats_output,
                thumbnail_output
            ]
        )

    return app

# -----------------------------------
# 7. LAUNCH APPLICATION
# -----------------------------------
if __name__ == "__main__":
    print("🚀 Starting YouTube Social Media Generator...")
    print("📱 Please wait while the application loads...")

    app = create_interface()
    # Using a general port and setting server_name='0.0.0.0' for deployment flexibility
    app.launch(
        share=True,
        debug=False,
        server_name="0.0.0.0",
        server_port=7860,
        show_error=True
    )
```
